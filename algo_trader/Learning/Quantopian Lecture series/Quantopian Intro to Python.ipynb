```python
def sum_values(*args): # * denotes an arbitary number of arguments
    sum_val = 0
    for i in args:
        sum_val += i
    return sum_val
```

```python
print(sum_values(1, 2, 3))
print(sum_values(10, 20, 30, 40, 50))
print(sum_values(4, 2, 5, 1, 10, 249, 25, 24, 13, 6, 4))
```

```python
def test_args(*args):
    print(type(args))

test_args(1, 2, 3, 4, 5, 6)
```

```python
def test_args(*args):
    type_list = []
    for i in args:
        type_list.append(type(i))
    return type_list

for a in test_args(1, 2, 3, 4, 5, 6):
    print(a)

print(test_args(1, 2, 3, 4, 5, 6))
```
