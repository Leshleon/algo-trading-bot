# 1. Means

### Arithmetic Mean

```python
x1 = [1, 2, 2, 3, 4, 5, 5, 7]
x2 = x1 + [100]

print("Mean of Z1:", sum(x1), "/", len(x1), "=", np.mean(x1))
print("Mean of Z1:", sum(x2), "/", len(x2), "=", np.mean(x2))
```

### Median

```python
print("Median of x1:", np.median(x1))
print("Median of x2:", np.median(x2))
```

### Mode

```python
print("One mode of x1:", stats.mode(x1)[0])
```

```python
def mode(l):
    counts = {}
    modes = set()
    for e in l:
        if e in counts:
            counts[e] += 1
        else:
            counts[e] = 1
    
    maxcount = max(list(counts.values()))
    modes = set()
    for e in counts:
        if counts[e] == maxcount:
            modes.add(e)
    
    return modes

mode(x1)       
```
